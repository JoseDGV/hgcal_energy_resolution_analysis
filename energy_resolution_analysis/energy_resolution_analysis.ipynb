```python
#define energy resolution fitting function 
def energy_resolution_func(E, p0, p1):
    return np.sqrt(p0 + p1/E)
```

```python
#define chi squared with error function
def chi2(obs, exp, error):
    return np.sum((obs - exp) ** 2 / (error ** 2))
```

```python
# read our results as a pandas dataframe
df = pd.read_csv('energy_resolution_'+particle_type+'.csv')
df
```

```python
#fit energy resolution function
initialParameters = np.array([1.0, 1.0])
params, pcov = scipy.optimize.curve_fit(energy_resolution_func, df['E'], df['std(E)/E'])
print(params, pcov)
```

```python
#get std deviation error on the parameters
params_err = np.sqrt(np.diag(pcov))
params_err
```

```python
#calculate the chi squared value taking the errors per point into account
predicted = energy_resolution_func(df['E'], params[0], params[1]) 
chisquared = np.round(chi2(predicted, df['std(E)/E'], df['std(E)/E']*1/np.sqrt(1000)),2)
chisquared
```
