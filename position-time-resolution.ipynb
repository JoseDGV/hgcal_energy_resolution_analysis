```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
E = [2, 3, 5, 10, 25, 50, 75, 100, 200]
positionX = [0.13, 0.12, 0.11, 0.1, 0.09, 0.08,0.07, 0.06, 0.05]
positionY = [0.14, 0.13, 0.11, 0.1, 0.09, 0.08, 0.07, 0.07, 0.06]
time = [0.04, 0.037, 0.036, 0.035, 0.034, 0.0329, 0.0327, 0.0349, 0.0327]

E_2 = [25, 50, 75, 100, 200]
events_all = [1792572, 4156791, 5473490,  5964720, 10461349 ]
events_filter = [640977, 1532609, 2059243, 2276187, 4101356 ]
```

```python
df_position = pd.DataFrame({'energy': E, 'positionX': positionX})
```

```python
# Plot the arrays
ax = df_position.plot.scatter(
    x='energy', 
    y='positionX', 
    logx=True,
    logy=False,
    title = 'Position Resolution - Photons',
    ylabel = '$σ_X$', 
    xlabel =  'Energy [GeV]',
    style='x',
    legend = None,
    zorder=10)

ax.set_axisbelow(True)
ax.yaxis.grid(color='gray', linestyle='dashed')
ax.xaxis.grid(color='gray', linestyle='dashed', which="both")
#ax.set_ylim([0., 0.15])

# Show the plot
ax.get_figure().savefig('position_resolutionY.png', bbox_inches = 'tight')
plt.show()
```

```python
df_time = pd.DataFrame({'energy': E, 'time': time})
```

```python
# Plot the arrays
ax = df_time.plot.scatter(
    x='energy', 
    y='time', 
    logx=True,
    logy=False,
    title = 'Time Resolution - Photons',
    ylabel = '$σ_t$', 
    xlabel =  'Energy [GeV]',
    style='x',
    legend = None,
    zorder=10)

ax.set_axisbelow(True)
ax.yaxis.grid(color='gray', linestyle='dashed')
ax.xaxis.grid(color='gray', linestyle='dashed', which="both")
#ax.set_ylim([0., 0.15])

# Show the plot
ax.get_figure().savefig('time_resolution.png', bbox_inches = 'tight')
plt.show()
```

```python
df_event_all = pd.DataFrame({'energy': E_2, 'events_all': events_all, 'events_filter': events_filter}).set_index('energy')
```

```python
# Plot all columns
ax = df_event_all.plot()
plt.title('# of events per energy level')
plt.xlabel('Energy')
plt.ylabel('# of Events')

ax.set_axisbelow(True)
ax.yaxis.grid(color='gray', linestyle='dashed')
ax.xaxis.grid(color='gray', linestyle='dashed', which="both")
plt.show()
```
